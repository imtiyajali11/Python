```python
# 4. Create a dictionary of 5 products and their prices. 
# Find the total price of all products using a loop.

product = { 
                'Mango': 1,
                'Banana': 2,
                'Grapes': 3,
                'Apple': 4,
                'Orange': 5
                }
type(product)
```

```python

print(product.keys())
print(product.values())
```

```python
total = 0

for i in product:
    total = product[i] + total
print(total)
```
