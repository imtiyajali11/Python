# common and uncommon from two lists

```python
# common from 2 list

a = [1,2,3,4,5,6,7,8]
b = [5,6,7,8,9,10,11]
```

```python
type(a),type(b)
```

```python
c = []
d = []
for i in range(len(a)):
    found = False
    for j in range(len(b)):
        if a[i] == b[j]:
            found = True
            c.append(a[i])
    if found == False:
        d.append(a[i])

for j in range(len(b)):
    found = False
    for i in range (len(a)):
        if b[j] == a[i]:
            found = True
            
    if found == False:
        d.append(b[j])

            
        
print('common:', c)
        
print('uncommon:',d)
```
