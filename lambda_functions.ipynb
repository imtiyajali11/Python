```python
# find out the odd and even of the given number suing user defined function

a = 12

if a% 2 == 0:
    print('Even')
else:
    print('Odd')
```

```python
def odeven(a):
    if a% 2 == 0:
        print('even')
    else:
        print('odd')
```

```python
odeven(3413)
```

```python
z = lambda a: "Even" if a%2 == 0 else "Odd"     # anonimous function
z(343)
```

```python
z(1244)
```
