# conditional statements 

## If Syntax

if condition : 

    code
else:
    code

```python
age  = 17
if age >= 18:               # if is keyword, age >= 18 is condition
	print('Can vote')       # if condition is true then print statement will be executed
else:                       # else is keyword, if condition is false then print statement will be executed
	print('Not Eligible').  # --- IGNORE ---
```

```python
percentage = 34
if percentage >= 35:
    print('Pass')
else:
    print('Fail')
```

```python
# no. = 0

no = 0
if no > 0:
    print('Positive')
elif no == 0:
    print('Zero')
else:
    print('Negative')
```

```python
a = 5
b = 3


if a > b:
    print('a is greater than b')
elif a == b:
    print('a is equal to b')
else:
    print('a is less than b')
```

```python
# max of 3 numbers

a = 8 
b = 3
c = 7

if a> b and a> c:
    print ('a is greater than b and c')
elif b > a and b > c:
    print('b is greater than a and c')
else:
    print('c is greater than a and b')  
```

```python
# grading system

per = 900

# a++ = 90
# a = 80
# b = 60
# c = 50
# d = 35
# Fail

if per <=100 and per >= 0:
    if per > 90:
        print ('Grade A++')
    elif per > 80:
        print('Grade A')
    elif per > 60:
        print('Grade B')
    elif per > 50:
        print('Grade C')
    elif per > 35:
        print('Grade D')
    else:
        print('Fail')
else: 
    print('Invalid Percentage')
```

#  salary and bonus
## salary upto 30k bonus 15%
## salary upto 50k bonus 10%
## salary upto 50k bonus 5%

    

```python
salary = 50000
if salary >= 0:
    if salary <= 30000:
        print('bonus 15%:', salary*0.15)
    elif salary <= 50000:
        print('bonus 10%', salary*0.10)
    else:
        print('bonus 5%:',salary*0.5)
else:
    print('invalid salary')
```
