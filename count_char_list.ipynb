```python
# lets count (STR)string elements in the list
# a = [200,'a', 12, 'b', 13,'ali', 14,'suraj',21]
i = 0 
count = 0
lst = []
cnt = 0

while i < len(a):
    if type(a[i]) == str:  
        lst += a[i]
        print(i, a[i])
        count += 1
        cnt += len(a[i]) 
        
    i += 1
print(lst, count, cnt)
```

```python
# lets count (int)string elements in the list
# a = [200,'a', 12, 'b', 13,'ali', 14,'suraj',21]
i = 0 
cnt = 0

while i < len(a):                   # loop range from 0 to 10
    if type(a[i]) == int:           # count only if integer in the list
        
        print(i, a[i])              # print index and value
        count += 1                  # i index increase by to go to next value in list
        
        
    i += 1
print(lst, count, cnt)
```
