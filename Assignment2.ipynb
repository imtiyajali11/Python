```python
a = [1,3,54,3,3,5,6,7]
#a = [1,-3,54,-3,3,5,6,0]
# i, j, k index
# a[i]+ a[j]+a[k] = 0
# i!=j!=k
a

n = len(a)
n

for i in range(n):
    for j in range(i+1, n):
        for k in range(j+1, n):
            if a[i]+a[j]+a[k] == 0:
                print('index:', i, j, k)
                print('Values:', a[i], a[j],a[k])

```

```python
a = [1,-3,54,-3,3,5,6,0]

n = len(a)


# i != j and  j != k and i != k

for i in range(n):
    for j in range(i+1, n):
        for k in range(j+1, n):
            if a[i]+a[j]+a[k] == 0 and a[i] != a[j] and  a[j] != a[k] and a[i] != a[k]:
                print('Index:',i,j,k)
                print('Values',a[i], a[j], a[k])
```
