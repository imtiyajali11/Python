```python
# 5. Find the largest number in a list using a loop.
a = [1,2,3,4,5, 12,12,15,17]
lst = a[0]
n = len(a)


for i in range(n):
    for j in range(i+1, n):
        if a[j] > a[i]:
            
            lst = (a[j])
print('Largest No.:', lst)

```
