```python
# 1. Create a tuple of 5 numbers and print each element using a loop.

a = (1,2,3,4)

for i in a:
    print(i)
```

```python
# 2. Find the sum of all elements in a tuple.

a = (1,2,3,4)
total = 0

for i in a:
    total = total + i 
print(total)

```

```python
def tupple_total_sum(a):
    total = 0

    for i in a:
        total = total + i 
    print(total)
```

```python
tupple_total_sum ( a = (1,2,3,4,5,6,7,8,9))
```

```python
# 3. Count how many times a specific number appears in a tuple using a loop.

a = (1,2,3,4,2)
type(a)
find = 2
count =0

for i in a:
    if find == i:
        count += 1
print('looking for:',find, '&', 'it appeared for:' ,count, 'times')

```

```python
def look_for_howManyTimes(find, list):

  
    count =0

    for i in list:
        if i == find:
            count += 1
    print('looking for:',find, '&', 'it appeared:' ,count, 'times')
```

```python
look_for_howManyTimes(4,(1,2,3,4,5,6,7,8,9,9,4,4,4))
```

```python
# 5. Find the smallest value in a tuple using a loop.
a = (1,2,3,4,56,-11, -100)
type(a)
```

```python
b = a[0]

for i in a:
    if i < b:
        b = i
print(b)
```

```python
def smallest_from_tupple(a):
    b = a[0]

    for i in a:
        if i < b:
            b = i
    print(b)
```
