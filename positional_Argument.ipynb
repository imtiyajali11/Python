# Positional Argument

```python
def info(first_name, last_name):
    return (f""" the first name of the person is {first_name}
and the last name is {last_name}""")
```

```python
info('Sawant', 'Rao') # 
```

```python
info(last_name= "Sawant", first_name="Om")
```

```python
# Default parameter

def info(first_name, last_name, batch = "M373"):
    print(f"""the first name of the person is {first_name}
and the last name is {last_name } and studies in batch {batch}""")
```

```python
info(last_name="Suraj", first_name= "Singh")
```
