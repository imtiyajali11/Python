```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import csv
import time
```

```python
search_item = "mobiles"
pages = 5

```

```python
headers = {
    "User-Agent":
    "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/137.0 Safari/537.36"
}
```

```python
titles = []
prices = []
ratings = []
rows = []
```

```python
for page in range(1, pages + 1):

    url = f"https://www.flipkart.com/search?q={search_item}&page={page}"

    print(f"Scraping Page {page}")

    response = requests.get(url, headers=headers)

    soup = BeautifulSoup(response.content, "html.parser")
```

```python
title = soup.find_all("div", class_="RG5Slk")

for item in title:
    titles.append(item.get_text(strip=True))
```

```python
products_data = []

for product in products:

    # Title
    title = product.find("div", class_="RG5Slk")
    title = title.get_text(strip=True) if title else ""

    # Price
    price = product.find("div", class_="hZ3P6w DeU9vF")
    price = price.get_text(strip=True) if price else ""

    # Rating
    rating = product.find("div", class_="MKiFS6")
    rating = rating.get_text(strip=True) if rating else ""

    # Specifications
    specs = product.find("ul", class_="HwRTzP")

    ram = display = camera = battery = processor = warranty = ""

    if specs:
        spec_list = [li.get_text(strip=True) for li in specs.find_all("li")]

        if len(spec_list) > 0:
            ram = spec_list[0]
        if len(spec_list) > 1:
            display = spec_list[1]
        if len(spec_list) > 2:
            camera = spec_list[2]
        if len(spec_list) > 3:
            battery = spec_list[3]
        if len(spec_list) > 4:
            processor = spec_list[4]
        if len(spec_list) > 5:
            warranty = spec_list[5]

    products_data.append({
        "Title": title,
        "Price": price,
        "Rating": rating,
        "RAM": ram,
        "Display": display,
        "Camera": camera,
        "Battery": battery,
        "Processor": processor,
        "Warranty": warranty
    })
```
