# Amazon mobile name web scraping


```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from urllib.parse import quote_plus
```

```python
# Amazon India search URL for mobile phones
search_text = "mobiles"
source_url = f"https://www.amazon.in/s?k={quote_plus(search_text)}"
source_url
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/124.0.0.0 Safari/537.36",
    "Accept-Language": "en-IN,en;q=0.9",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    "Referer": "https://www.amazon.in/",
    "Connection": "keep-alive",
}
```

```python
response = requests.get(source_url, headers=headers, timeout=2)
```

```python
print("Status code:", response.status_code)
print("Final URL:", response.url)
```

```python
if response.status_code == 200:
    parse = BeautifulSoup(response.text, "html.parser")
    print("Page loaded successfully")
elif response.status_code == 503:
    print("Amazon returned 503. It may be blocking automated scraping.")
    print("Try again later, use slower requests, or use an approved API.")
else:
    response.raise_for_status()
```

```python
parse = BeautifulSoup(response.text, "html.parser")
```

```python
# Each product in Amazon search results is usually inside this result card.
product_cards = parse.select('div.s-main-slot div[data-component-type="s-search-result"]')
len(product_cards)
```

```python
mobile_names = []

for card in product_cards:
    title_tag = card.select_one('h2 span')
    if title_tag:
        mobile_names.append(title_tag.get_text(strip=True))

mobile_names[:10]
```

```python
mobile_df = pd.DataFrame({"mobile_name": mobile_names})
mobile_df.head()
```

```python
mobile_df.to_csv("amazon_mobile_names.csv", index=False)
print(f"Saved {len(mobile_df)} mobile names to amazon_mobile_names.csv")
```

Note: Amazon often changes its HTML and may show captcha/robot-check pages. For real projects, check the website terms and scrape slowly, or use an official/approved API when available.
