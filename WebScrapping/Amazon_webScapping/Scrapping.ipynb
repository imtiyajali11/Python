```python
data = []

for product in product_cards:
    name_tag = product.select_one("h2 span")
    price_tag = product.select_one("span.a-price-whole")
    rating_tag = product.select_one("span.a-icon-alt")
    link_tag = product.select_one("h2 a")

    name = name_tag.get_text(strip=True) if name_tag else None
    price = price_tag.get_text(strip=True) if price_tag else None
    rating = rating_tag.get_text(strip=True) if rating_tag else None
    link = "https://www.amazon.in" + link_tag["href"] if link_tag else None

    if name:
        data.append({
            "mobile_name": name,
            "price": price,
            "rating": rating,
            "link": link
        })

df = pd.DataFrame(data)
df
```

```python
df.to_csv("amazon_mobiles.csv", index=False)

print("Data saved successfully")
```
