```python
url = "https://quotes.toscrape.com/"
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import csv
```

```python
headers = {
'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}
```

```python
response = requests.get(url, headers = headers)
response
```

```python
soup = BeautifulSoup(response.content, "html.parser")
soup
```

```python
quotes = soup.find_all('div', class_='quote')
quotes
```

```python

# now let's create a single list table which will contain all quotes, author, and tags related to it.
```

```python


data = []                                                               ## to save all data in single list

for q in quotes:
    quote = q.find('span', class_='text').get_text(strip = True)          ## This will get all the authors

    author = q.find('small', class_= 'author').get_text(strip = True)      ## this will get all the authors

    tags = q.find_all('a', class_= 'tag')                ## this will get all the tags 
    t = []

    for i in tags:
        t.append(i.get_text(strip = True))

    
    data.append({
                'quote' : quote,
                'author': author,
                'tags': '. '.join(t)
                })
    
    


data   
```

## # save file to csv using pd

```python


import pandas as pd
```

```python
df = pd.DataFrame(data)
```

```python
df.to_csv('quotes_data.csv', index = 'False')
print('Data saved successfully')
print(df)
```
