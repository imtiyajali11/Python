# Country data scraping practice


```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://www.scrapethissite.com/pages/simple/"
response = requests.get(url, timeout=15)
print(response.status_code)
response.raise_for_status()
```

```python
soup = BeautifulSoup(response.content, "html.parser")
```

```python
page_title = soup.title.get_text(strip=True)
page_heading = soup.find("h1").get_text(strip=True)
total_items = soup.find("small").get_text(strip=True)

print(page_title)
print(page_heading)
print(total_items)
```

```python
description = soup.find("p", class_="lead").get_text(strip=True)
formatted_description = description.replace(". ", ".\n")
print(formatted_description)
```

```python
countries = soup.find_all("div", class_="country")
print(len(countries))
```

```python
first_country = countries[0]
country_name = first_country.find("h3", class_="country-name").get_text(strip=True)
country_capital = first_country.find("span", class_="country-capital").get_text(strip=True)
country_population = first_country.find("span", class_="country-population").get_text(strip=True)
country_area = first_country.find("span", class_="country-area").get_text(strip=True)

print(country_name, country_capital, country_population, country_area)
```

```python
data = []

for country in countries:
    country_name = country.find("h3", class_="country-name").get_text(strip=True)
    country_capital = country.find("span", class_="country-capital").get_text(strip=True)
    country_population = country.find("span", class_="country-population").get_text(strip=True)
    country_area = country.find("span", class_="country-area").get_text(strip=True)

    data.append({
        "country_name": country_name,
        "country_capital": country_capital,
        "country_population": country_population,
        "country_area": country_area,
    })

data[:5]
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
output_path = "countries_data.csv"
df.to_csv(output_path, index=False)
print(f"Data saved successfully to {output_path}")
```
