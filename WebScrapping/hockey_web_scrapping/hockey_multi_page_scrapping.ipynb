# Scrapping hockey data for MultiPage

```python
# Import libraies

import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
```

```python

def Hockey_team_dataScrapping(pages):

    url = f"https://www.scrapethissite.com/pages/forms/?page_num={page}&per_page=25"

    headers = {
    'User-Agent': 'Mozilla/5.0'
    }


    response = requests.get(url, headers= headers)      # this checks the page response if 200, 
                                                            #then ok or else page load is failed
                                                            

    soup = BeautifulSoup(response.content, 'html.parser')   # it will download all the pages


    team_table = soup.find_all("tr", class_="team")


    print(f' {page} scrapped from {url}  - rows: {len(team_table)}')

    team_data = []

    for data in team_table:
        name = data.find('td', class_ = 'name').get_text(strip=True)
        year = data.find('td', class_ = 'year').get_text(strip=True)
        wins = data.find('td', class_ = 'wins').get_text(strip=True)
        losses = data.find('td', class_ = 'losses').get_text(strip=True)
        ot_losses = data.find('td', class_ = 'ot-losses').get_text(strip=True)
        win_percentage = data.find('td', class_ = 'pct').get_text(strip=True)
        gf = data.find('td', class_ = 'gf').get_text(strip=True)
        ga = data.find('td', class_ = 'ga').get_text(strip=True)
        win_loss = data.find('td', class_ = 'diff').get_text(strip=True)


        team_data.append({
                    'Team name' : name,
                    'Year' : year,
                    'Wins' : wins,
                    'Losses': losses,
                    'OT Losses': ot_losses,
                    'Win %': win_percentage,
                    'Goals For (GF)': gf,
                    'Goals Against (GA)': ga,
                    '+/-': win_loss
                        })

    return team_data



```

```python
all_scrapped_data = []

pages = 25

for p in range(1, pages + 1):
    page_data = Hockey_team_dataScrapping(p)

    if page_data:
        all_scrapped_data.extend(page_data)

    time.sleep(0.1)

df = pd.DataFrame(all_scrapped_data)

df.to_csv("team_hockey_data.csv", index=False)

print("Data saved successfully.")
print(f"Total rows scraped: {len(df)}")
```
