```python
# 1. Create a tuple of 5 numbers and print each element using a loop.

a = (1, 2, 3, 4, 5)
type(a)
i =0

while i in range(i,len(a),1):
    print(a[i])
    i += 1
```

```python
# 2. Find the sum of all elements in a tuple.

a = (1, 2, 3, 4, 5)
total = 0
i = 0
while i in range(len(a)):
    total = sum(a)
    i += 1
print(total)
```

```python
# 4. Print only the odd numbers from a tuple.

a = (1, 2, 3, 4, 5, 2, 2)
for i in a:
    if i%2 != 0:
        print(i)
```

```python
# 5. Find the smallest value in a tuple using a loop.

a = (1100, 1122, 1343, 12344, 12345,3562, 572)

b = a[0]

for i in a:
    if i < b:
        b = i

print(b)
```
