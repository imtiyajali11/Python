```python
# print 10 to 0 numbers

x = 10
while x >= 1:
    x = x - 1
    print(x) 
```

```python
# even from 1 to 10 using while loop

i = 1

while i < 10:

    if i % 2 == 0:
    #    print('Even Numbers')
        print(i)

    
    else:
        print(i)

    i += 1
    
```
