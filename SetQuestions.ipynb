```python
# 1. Create a set of numbers and print all elements using a loop.

a = {1,2,3,4,5,6,7,8}
type(a)
```

```python
a = {1,2,3,4,5,6,7,8}

for i in a:
    print(i)
```

```python
a = {1,2,3,4,5,6,7,8}
a_list = list(a)
i = 0
while i in range(0, len(a_list)):
    print(a_list[i])
    i += 1
```

```python
# 2. Count how many elements are present in a set using a loop.
a = {1,2,3,4,5,6,7,8}
b = list(a)
count = 0 

for i in b:
    count += 1
print(count)
```

```python
a = {1,2,3,4,5,6,7,8}
count = 0

for i in a:
    count += 1
print(count)
```

```python
# 3. Print only numbers greater than 20 from a set.

a = {1,2,3,4,5,6,7,8,21}

for i in a:
    if i > 20:
        print(i)
```

```python
# 4. Find the sum of all values in a set using a loop.

a = {1,2,3,4,5,6,7,8,21}
total = 0
for i in a:
    total = total + i
    print(total)
print(total)
```

```python
# 5. Print the square of each element in a set.

a = {1,2,3,4,5,6,7,8,21}
total = 0

for i in a:
    print(i**2)
```
