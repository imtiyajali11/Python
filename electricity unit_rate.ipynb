# Unit rate
### upto 100 ₹5/unit
### 101 - 200 ₹7/unit
### above 200 ₹10/unit

```python
# input
unit = 300
```

```python


# If unit is less than zero condition
if unit > 0:
    if unit <=100:
        electricity_Bill = unit * 5
        print('Your electricity bill:', electricity_Bill )
    elif unit <= 200:
        electricity_Bill = unit * 7
        print('Your electricity bill:', electricity_Bill )
    elif unit > 200:
        electricity_Bill = unit * 10
        print('Your electricity bill:', electricity_Bill )
else:
    print('input correct unit between 1-999')
```
