```python
# a = [1,3,54,3,3,5,6,7]
a = [1,-3,54,-3,3,5,6,0]
# i, j, k index
# a[i]+ a[j]+a[k] = 0
# i!=j!=k
a

n = len(a)
n
count = 1

for i in range(n):
    for j in range(i+1, n):
        for k in range(j+1, n):
            if a[i]+a[j]+a[k] == 0:
                print(count)
                print('index:', i, j, k)
                print('Values:', a[i], a[j],a[k])
                count +=1


```
