```python
#1. Count how many vowels are present in a list of characters.
# , 'manoj', 'Rahul'
a = ['ali', 'eli', 'rahul','e']
count = 0

for i in a:
    for j in i:
        if j in 'aeiouAEIOU':
            count += 1
    
print(count)
```

```python
#2. Find the average of numbers stored in a tuple.

a = (1,2,3,4,5,9)

for i in a:
    avg = int(sum(a)/len(a))
print(avg)
```

```python
#5. Find the longest word from a list of words using a loop.


a = ['ali','rahul','VenugopalIyer']
long = ''

for i in a:
    if len(i) > len(long):
        long = i
print(long)
```
