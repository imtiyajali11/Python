```python
# print vowels from above variable

i = 0 
l = len(name)
while i < l:
    if name[i] in "aeiouAEIOU":
        print(name[i])
    i += 1
```

```python
i = 0 
l = len(name)
c = 0

# print count of consonants
while i < l:
    if name[i] not in "aeiouAEIOU":

        c = c +1
        print(name[i])
        print(c)
    i += 1
    
```
