```python
# print 1 to 10 no. using for loop

for i in range(1, 11, 2):
    print(i)
```

```python
# print 10 to -10 no using for loop
for i in range(10,-11,-1):
    print(i)
```

```python
# print series -9 to 9

for i in range (-9, 11,3):
    print(i)
```
