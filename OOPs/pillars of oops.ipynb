```python
class animal: 
    def sound(self):
        print(f'This animal makes sound')



class pet:
    def cat(self):

        print('it comes under top inhouse animal and it makes sound "Meow"')
```

```python
ani = animal ()
print(ani.sound())
catt = pet()
print(catt.cat())
```

```python
class dog(animal):
 #   pass                     pass can be used to pass and execute the code or just to execute code if there is any.
    def bark(self):
        print('dog can bark')

d = dog()
print(d.sound())
print(d.bark())
```

```python
class cat(pet):
    def meow(self):
        print('can can meow')


c = cat()
print(c.meow())
```

# 3. Polymorphism

Polymorphism (from Greek "many-shaped") refers to the ability
of a single entity, message, or symbol to take on different 
forms or behave in various ways depending on the context.

```python
class creditcard:
    def pay(self):
        print('paid using credit card')

class upi:
    def pay(self):
        print('paid using UPI')

class cash:
    def pay(self):
        print('paid using cash')



payments = [creditcard(), upi(), cash()]

for i in payments:
    i.pay()
```

```python
# 5. Data Abstraction


from abc import ABC, abstractmethod

class shape(ABC):
    @abstractmethod
    def area(self):
        pass


class square(shape):
    def __init__(self,side):
        self.side = side
    def area(self):
        print(self.side**2)
    def l(self):
        print(f'The square of {self.side} : {self.side**2}')

s = square(5)
s.area()
s.l()
```
