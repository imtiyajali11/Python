# Magic Methods (Dunder Methods) 
Dunder (Double Under) methods customize object behavior.

```python
class Book:
        def __init__(self, title, pages):
                self.title = title 
 
                self.pages = pages
        def __str__(self):         
                return f"Book: {self.title}, Pages: {self.pages}"      
                
        def __len__(self):         
                return self.pages 
book = Book("Python Mastery", 450) 
print(book)  # Calls __str__() print(len(book))  # Calls __len__()
```
