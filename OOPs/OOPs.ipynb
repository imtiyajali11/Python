# init constructor : to initialize object data

```python
# init constructor : to initialize object data
# self keyword to pass extra data or objects (it acts as a medium to transfer data to pass on to next variable)

class student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

```python
s1 = student('ali', 62)
s2 = student('william', 31)
s3 = student('peter', 18)
```

```python
print(s1.name)
print(s2.name)
```

# Methods

```python
# Methods

class subway:
    def __init__(self, key, points, box, jump):

        self.key = key
        self.points = points
        self.box = box
        self.jump = jump


    def rewards(self):                      # method: function in class, features of class.
        print(f'You have {self.points} reward points')
```

```python
g1 = subway('5', 100, 'green','high')
```

```python
print(f'{g1.key}')
```

```python
g1.rewards()
```
