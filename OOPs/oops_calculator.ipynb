```python
# conctrucotr
class calculator:
    def __init__(self, a, b,operator):
        self.a = a
        self.b = b
        self.operator = operator
    
# method
    def calculate (self):
        if self.operator == '+':
            print (self.a+self.b)
        elif self.operator == '-':
            print(self.a-self.b)
        elif self.operator == '*':
            print(self.a* self.b)
        elif self.operator == '/':
            print(self.a/self.b)

```

```python
c = calculator(1,2,'+')
c.calculate()
```

```python
print(f'{c.calculate}')
```
