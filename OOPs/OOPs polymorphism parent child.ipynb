# polymorphism

It allows metholds to be redefined in child class

```python
class Bird:

    def sound(self):
        return "some generic sound"
    

class Sparrow(Bird):
    def sound(self):
        return"Chirp Chirp"
    
class Crow(Bird):
    def sound(self):
        return "Caw Caw"
    
birds =[Sparrow(), Crow()]

for bird in birds:
    print(bird.sound())
```
