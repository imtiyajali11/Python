```python
def calculator(a, b, operator):


    try:


        if operator == '+':
            print (a+b)
        elif operator == '-':
            print(a-b)
        elif operator == '*':
            print(* b)
        elif operator == '/':
            print(a/b)



    
    except NameError: 
            print('Got an name error')
    except exception as e:
            print('You got an error')
```
