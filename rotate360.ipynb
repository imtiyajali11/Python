```python
n = 5
l = [1,2,3,4,5,6,7,8,9,0]
le = len(l)
i = 0
result = []
# rotate and whole numbers from the index (n) from end to start and start will move to end

while n < le:
    result = l[n:]+ l[:n]
#    result.append(l[n])
    n += 1
    i += 1

    print(result)
```

```python
n = 5
l = [1,2,3,4,5,6,7,8,9,0]
result = []

for i in range(n, len(l)):
    result.append(l[i])
    print("adding from end part:", result)

for j in range(n):
    result.append(l[j])
    print("adding from start part:", result)

print("final:", result)
```
