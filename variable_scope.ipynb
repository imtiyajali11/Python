```python
# variable scope
# global
# local
```

```python
a = 11 # global variable

def  add():
    a = 20
    b = 1    # local variable
    return (a+b)

add()
```

```python
add()
```

```python
a = 11 # global variable

def  add():
    global a             # declare a as global and it will be 
                         #updated in outside local as well
    a = 20
    b = 1    # local variable
    return (a+b)

add()
```
