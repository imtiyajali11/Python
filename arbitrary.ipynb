```python
# Default parameter

def info(*args):

    fname = args[0]
    lname = args[1]
    batch = args[2]
    rollno = args [3]
    print(f"""the first name of the person is {fname} 
 and the last name is {lname } and studies in batch {batch} 
 and roll no is {rollno}""")
```

```python
info('Om',"Sanjay", "M373", "11")
```

```python
# Default parameter

def info(*kargs):
    fname = kargs['first_name']
    lname = kargs['last_name']
    print(f"""the first name of the person is {fname} 
 and the last name is {lname }""")
```

```python
info(first_name = "Rohit", last_name= "Sharma")
```
