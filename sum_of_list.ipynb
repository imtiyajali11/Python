```python
# a = [11, -12, 11, -1, 12]       list

i     = 0    # iterator
total = 0    # running total for + values
total1 = 0   # running total for -ve values

while i < len(a):              # running loop for length of list
    if a[i] >= 0:              # checking +ve values 
        total = total + a[i]   # adding to running total 
      #  print(total)
    else:
        total1 = total1 + a[i]
       # print(total1)
    i += 1
print('total of +ve values: ',total,'      total of +ve values: ',total1,'     length of list:', len(a)) 
```
